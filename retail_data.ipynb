# Month dataset clean

```python
# order_date change to datetime
df['order_date']=pd.to_datetime(df['order_date'])
```

```python
df.dtypes
```

```python
df['month_name']=df['order_date'].dt.month_name()

df['month_num']=df['order_date'].dt.month

df['quarter']=df['order_date'].dt.quarter

df['year']=df['order_date'].dt.year
```

```python
df
```

# measure data

```python
df.columns
```

```python
# Total customer

total_customer =df['customer_name'].nunique()

print("Total Customer : ",total_customer)
```

```python
# Total sales

total_sales=round(df['sales'].sum(),2)

print("Total sales :",total_sales)
```

```python
# Average Profit

avg_profit = round(df['profit'].mean(),2)

print('averge profit : ', avg_profit)
```

```python
# Region wise sales

region_wise_sales=df.groupby('region')['sales'].sum()

print('Region wise sales: ', region_wise_sales)

```

```python
# Top 5 Product

df['product'].unique()

print('Top 5 product')

top_5_product=(df.groupby('product')['sales'].sum().sort_values(ascending=False).head(5))

# top_5_product=(df.groupby(['category','product'])['sales'].sum().sort_values(ascending=False).head(5))

print('Top 5 product : ',top_5_product)
```

```python
# city by sales

city_by_sales = df.groupby('city')['sales'].sum().sort_values(ascending=False)

print('city by sales : ',city_by_sales)
```

```python
# high and low sales in month

high_sales_month=df.groupby('month_name')['sales'].sum().sort_values(ascending=False).head(5)
print('high_sales_month : ', high_sales_month)

print('\n')

low_sales_month=df.groupby('month_name')['sales'].sum().sort_values(ascending=True).head(5)
print('low_sales_month : ', low_sales_month)
```

```python
# Product quantity count

product_quantity_count=df.groupby('product')['quantity'].count().sort_values(ascending=False)

print('Product quantity count : ',product_quantity_count)
```

```python
# as_index=True   → grouped column becomes index
# as_index=False  → grouped column stays as column

sales_month = (
    df.groupby(
        ['month_num', 'month_name'],
        as_index=False
    )['sales']
    .sum()
    .sort_values('month_num')
)
print('sales_month : ', sales_month)

fig=px.line(sales_month,x='month_name',y='sales',title='monthly sales',labels={'month_name':'Month','sales':'Total Sales'},color_discrete_sequence=['#5d9854'])

fig.show()
```

```python
import plotly.express as px

top5 = (
    df.groupby('product', as_index=False)['quantity']
      .sum()
      .sort_values('quantity', ascending=False)
      .head(5)
)

fig = px.bar(
    top5,
    x='product',
    y='quantity',
    title='Top 5 Products by Quantity',
    labels={
        'product': 'Product Name',
        'quantity': 'Total Quantity'
    },
    color_discrete_sequence=['blue']
)

fig.update_layout(
    plot_bgcolor='lightgray',
    paper_bgcolor='white',
    xaxis_title='Product Name',
    yaxis_title='Total Quantity'
)

fig.show()
```

```python
city_coordinates = {
    'Chennai': (13.0827, 80.2707),
    'Madurai': (9.9252, 78.1198),
    'Salem': (11.6643, 78.1460),
    'Trichy': (10.7905, 78.7047),
    'Coimbatore': (11.0168, 76.9558),
    'Tirunelveli': (8.7139, 77.7567)
}

df['latitude'] = df['city'].map(lambda x: city_coordinates[x][0])
df['longitude'] = df['city'].map(lambda x: city_coordinates[x][1])
```

```python
df
```

```python
city_sales=(df.groupby(['city','latitude','longitude'],as_index=False)['sales'].sum())


fig=px.scatter_map(city_sales,lat='latitude',lon='longitude',size='sales',color='sales',hover_name='city',title='city wise sales',zoom=5)

fig.show()
```

```python
category_sales = (
    df.groupby(['category', 'product'], as_index=False)['sales']
      .sum()
)

fig = px.sunburst(
    category_sales,
    path=['category', 'product'],
    values='sales',
    title='Sales Distribution by Category& Product'
)

fig.show()
```
